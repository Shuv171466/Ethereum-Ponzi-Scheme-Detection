```python
import pandas as pd
from decimal import Decimal, InvalidOperation

def safe_eth_to_wei(value):
    try:
        if pd.isna(value):
            return None
        
        value = str(value).strip()
        
        # Remove unwanted text if needed
        value = value.replace("ETH", "")
        
        if value == "":
            return None
        
        return Decimal(value) * Decimal("1e18")
    
    except (InvalidOperation, ValueError):
        return None

df["Amount"] = df["Amount"].apply(safe_eth_to_wei)
```

```python
df["Amount"] = df["Amount"].apply(
    lambda x: f"{x:.0E}" if x is not None else None
)
```
