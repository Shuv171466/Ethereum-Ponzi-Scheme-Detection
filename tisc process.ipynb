```python
import pandas as pd

# Load main transaction dataset
tx_df = pd.read_csv("Transaction.csv")

# Load internal transaction dataset
internal_df = pd.read_csv("Internal.csv")

# Count number of internal transactions per parent transaction hash
internal_count = internal_df.groupby("Parent Transaction Hash").size()

# Merge count into main dataframe as TISC column
tx_df["TISC"] = tx_df["Transaction Hash"].map(internal_count).fillna(0).astype(int)

# Save merged file
tx_df.to_csv("Transaction_with_TISC_791.csv", index=False)

print("Merged file created: Transaction_with_TISC.csv")
```
