```python
import pandas as pd

# Load CSV files
df1 = pd.read_csv("1.csv")
df2 = pd.read_csv("2.csv")
df3=pd.read_csv("3.csv")
df4 = pd.read_csv("4.csv")
df5 = pd.read_csv("5.csv")
df6=pd.read_csv("6.csv")

# Combine (stack rows)
combined = pd.concat([df1, df2,df3,df4, df5,df6], ignore_index=True)

# Save new CSV
combined.to_csv("Transaction_24516791.csv", index=False)

print("Files combined successfully!")
```
